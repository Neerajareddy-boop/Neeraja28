```python
def reverse_integer(x):
    INT_MIN = -(2**31)
    INT_MAX = 2**31 - 1

    sign = -1 if x < 0 else 1
    x = abs(x)
    reversed_number = 0

    while x > 0:
        digit = x % 10
        x //= 10

        # Check the limit before adding the next digit.
        if reversed_number > INT_MAX // 10 or (
            reversed_number == INT_MAX // 10 and digit > INT_MAX % 10
        ):
            return 0

        reversed_number = reversed_number * 10 + digit

    result = sign * reversed_number
    return result if INT_MIN <= result <= INT_MAX else 0


x = int(input("Enter a signed 32-bit integer: "))
print("Reversed integer:", reverse_integer(x))
```
